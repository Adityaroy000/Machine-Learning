# Decison Tree
## post pruning

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
df = sns.load_dataset('iris')
df.head()
```

```python
# independent and dependent feature
X = df.iloc[:,:-1]
y = iris.target
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.33,random_state=42)
X_train
```

```python
from sklearn.tree import DecisionTreeClassifier
treemodel = DecisionTreeClassifier()

treemodel.fit(X_train,y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(treemodel,filled=True)
```

```python
#post pruning till maxdepth = 2 bcz on level 2 green box has versicolor 34 and virginica 4 so after this there is no need to construct tree as it will give the answer whoever will be in majority same for right side blue color box also here verginica is 30 and versicolor is 1 
from sklearn.tree import DecisionTreeClassifier
treemodel2 = DecisionTreeClassifier(max_depth=2)

treemodel2.fit(X_train,y_train)

from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(treemodel2,filled=True)
```

```python
y_pred = treemodel2.predict(X_test)
```

```python
# Accuracy check
from sklearn.metrics import accuracy_score,classification_report
score = accuracy_score(y_pred,y_test)
print(score) #98% accuracy 
```

```python
print(classification_report(y_pred,y_test))
```

## pre prunning

```python
model = DecisionTreeClassifier()
parameter = {
    'criterion':['gini','entropy','log_loss'],
    'splitter' :['random','best'],
    'max_depth': [1,2,3,4,5],
    'max_features':['auto','sqrt','log2']
}
from sklearn.model_selection import GridSearchCV
cv = GridSearchCV(model,param_grid=parameter,cv = 5,scoring='accuracy')
```

```python
cv.fit(X_train,y_train)
```

```python
cv.best_params_
```

```python
y_preds = cv.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
score = accuracy_score(y_preds,y_test)
print(score)
```

```python
print(classification_report(y_preds,y_test))
```
